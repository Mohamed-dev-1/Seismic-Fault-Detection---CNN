## The big picture

**The goal :** build a classification model using CNN that looks at a seismic 2D image (a picture of the underground rock layers) and automatically answers **"Is there a fault(a break in the layers) in this section or not ?"**

**What does SEISMIC mean ? is it the same as Earthquake ?**

* The meaning of **Seismic** is related to vibrations or waves that travels through the earth.

* Earthquakes can create seismic waves (natural,uncontrolled,sudden).

* Humans can also create small seismic waves using: Vibrating trucks, air guns(underwater) or small explosive charges, and the main reason is to discover what's underground, and it is called **Seismic exploration**.

* So,all earthquakes are seismic events, but not all seismic waves come from earthquakes (could be Seismic explorations).


**Remember :** that our main goal is not studying or predicting earthquakes.

* **Our main goal is : Seismic exploration** (discover what is underground), ex : "What does the rock looks like a few kilometers under the area N°1 without digging that area?", **then train a CNN model that recognizes faults to decide is there a fault or not in that area**.


**What is a fault ?**

* Rock layers were formed in a flat and continuous, over millions of years.

* But the earth's crust is not static. tectonic stress can crack the rock and shift one side of the rock up or down or right or left relative to the other side.

* That crack-and-shift is a fault. When you look at the seismic image, instead of seeing smooth continuous layers, you see the layers break and offset at some point.


**Why doed finding faults is important ?**

* Oil/gas traps: faults can act like underground walls that trap oil/gas in a pocket, so finding a fault can mean finding a reservoir worth drilling. Other faults can let oil/gas leak away, so the type of fault matters commercially.

* Earthquake risk: faults are exactly the kind of structure that can slip suddenly and cause an earthquake, so knowing where they are matters for hazard assessment.

* Drilling safety: drilling through a fault zone can be unstable/dangerous for the drill equipment.

**Other reasons of Seismic exploration :**

    - Look for Oil and Gas reservoirs
    - Look for groundwater
    - Study geological hazards (fault lines that could cause earthquakes later)
    - Evaluating Construction sites before starting the constructing.

<h3><b> Where do Seismic exploration operates excatly within the crust ? </b></h3>

* Most oil and gas reservoirs sit roughly 1 to 5 kilometers below the surface(the top of the gound).

* Compare that to the crust itself, which is about 30 to 40 km thick under continents (much thinner under oceans, ~5-10 km)

* So Oil and Gas reservoirs are found in just the upper portion of the crust.


Now the question is : Ok, we knew that seismic exploration operates only on **the top part of the crust, where Oil and Gas reservoirs exists**. But in which layer of the top part exactly?

* We can not define a fixed specific layer, it's about finding the specific geological setup, which indicates that there is Oil and Gas in this area.

The setup is composed of 3 factors :

   1. **Source rock :** an organic-rich rock layer (usually shale) where, over millions of years under heat and pressure, dead organic material converted into oil and gas.

   2. **Migration :** oil and gas are lighter than the water that saturates rock pores, so they slowly migrate upward through permeable rock, trying to rise.

   3. **A trap :** this is the critical part. The oil/gas keep rising until they hit a barrier, an impermeable rock layer (like a dense shale or salt layer) that blocks further upward movement. This is called a cap rock or seal. Under that seal, the oil and gas accumulate inside a porous, permeable rock layer (commonly sandstone or limestone) called the reservoir rock.


   So, The Gas and Oil and Water are trapped on top with the **Cap Rock (Seal)** and at the buttom with the **Reservoi Rock(Porous)** in this order :

<h3><b>Seismic Exploration: Full Step-by-Step Process :</b></h3>

Context: All of this happens within the upper part of the earth's crust, typically the top 1 to 5 km, where sedimentary rock layers (sandstone, shale, limestone, etc.) are stacked. We never go deeper than the crust; the mantle and core are irrelevant here.

**Step 1 [Setup]: Source and Receivers**

Geophysical companies deploy two things over the target area:

* Source: trucks equipped with large vibrators (on land) or air guns towed by a boat (at sea). These generate controlled acoustic energy(sound waves) ,essentially artificial, repeatable sound pulses sent into the ground.

* Receivers: thousands of geophones (on land) or hydrophones (at sea), extremely sensitive instruments placed across the survey area, whose only job is to detect and record any sound that returns to the surface.

**Step 2 [Sending the Sound Waves]**

The vibrator trucks send thousands of sound pulses down into the ground, one survey position at a time. Each pulse travels downward through the crust as a wave of acoustic energy.

**Step 3 [Reflection at Rock Layer Boundaries]**

As the wave travels down, it passes through different sedimentary rock layers. Each time it crosses a boundary between two layers with different density/composition (for example, where shale meets sandstone, or where a cap rock/seal sits above a reservoir rock), part of the wave's energy reflects back upward, while the rest continues deeper to hit the next boundary. This repeats at every layer interface the wave encounters, including, if present, the sharp offset created by a fault.

**Step 4 [Recording the Echoes]**

The geophones/hydrophones at the surface pick up every reflected echo returning from underground, recording both when each echo arrives and how strong it is.

**Step 5 [Depth from Timing]**

This is the key physical principle: the longer an echo takes to return, the deeper the rock boundary that produced it. (Same logic as an echo in a canyon, a delayed echo means a farther wall.) So travel time directly encodes depth.

**Step 6 [Building the Image]**

Each geophone records one continuous signal over time, that single recording is called a trace.

Now line up many geophones along the survey line, and place their traces side by side. You get a 2D grid:

Horizontal axis = which geophone recorded it (i.e., position along the line)
Vertical axis = time since the pulse was sent (i.e., how deep, longer time means deeper)

Plotting the wave amplitude at every point as brightness/color produces a picture where dark and light bands mark exactly where the rock layer boundaries are, this is the seismic section. Extending this from a single line to a full grid of geophones produces a 3D seismic volume instead of just one 2D slice.

**Step 7 [What You're Actually Looking At]**

The resulting image is not an artistic reconstruction — it is a direct, physical map of subsurface layer boundaries: sedimentary rock layers, the source rock (organic-rich shale where oil/gas formed), the cap rock/seal (impermeable barrier), and the reservoir rock (porous layer where oil, gas, and water accumulate, stacked by density). A fault appears in this image as a visible break or sideways offset in these otherwise continuous layers,which is exactly the pattern our CNN will be trained to recognize.


So the process is simply :

- send controlled sound ,

- it reflects off real rock layer boundaries in the crust ,

- time and strength of each reflection is recorded ,

- stacked recordings form a picture of the actual sedimentary layers, seals, reservoirs, and faults underground.
