## The big picture

**The goal :** build a classification model using CNN that looks at a seismic 2D image (a picture of the underground rock layers) and automatically answers **"Is there a fault(a break in the layers) in this section or not ?"**

**What does SEISMIC mean ? is it the same as Earthquake ?**

* The meaning of **Seismic** is related to vibrations or waves that travels through the earth.

* Earthquakes can create seismic waves (natural,uncontrolled,sudden).

* Humans can also create small seismic waves using: Vibrating trucks, air guns(underwater) or small explosive charges, and the main reason is to discover what's underground, and it is called **Seismic exploration**.

* So,all earthquakes are seismic events, but not all seismic waves come from earthquakes (could be Seismic explorations).
